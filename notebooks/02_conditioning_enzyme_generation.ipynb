## **1.1. A bit of context about enzymes**

* Enzymes are **biological catalysts.**

* Enzymes are attractive because they can accelerate chemical reactions while working under very mild conditions (no need to use harsh chemicals, or high temperature and pressure). They are very attractive as green alternatives for industrial chemical processes.


**ENZYME STUCTURE:**
  * **Active site**: region where the substrate binds to produce the reaction. Higly specific for one substrate or a group of substrates.
  * **Cofactors and coenzymes**: additional molecules required to realize the  function:
    * Cofactors: inorganic ions (mg2+, Zn2+...)
    * Coenzymes: organic molecules (NAD+, FAD...)


**MECHANISM**:
* **Steps**:
1. Substrate binding: formation of enzyme-substrate complex.
2. Catalysis: Conversion of substrates into products.
3. Product Release: generation of the product.


enzymes.png

Their active sites are highly specialized regions that lower the activation energy, accelerating chemical transformations by several orders of magnitude

**Enzymes and activation energy**

Enzymes lower the activation energy by:
- Stabilizing the transition state
- Bringing reactacts closer together (proximity effect, increase of relative concentration)
- Creating a favorable electrostatic and acid-base environment

In the plot above we can see a schema of how an enzyme speeds up a given reaction time by lowering the activation energy of the process.

### Kinetics: why lowering the barrier matters

The overall free-energy change, $\Delta G_{\mathrm{rxn}}$, tells us whether reactants or products are thermodynamically favored, but it does not tell us how quickly the reaction occurs. The reaction rate is controlled by the activation free energy, $\Delta G^\ddagger$: the free-energy difference between the reactants and the transition state.

According to the transition-state theory, the rate and the activation free energy are related as:

$$
k = \frac{k_{\mathrm{B}}T}{h}
\exp\left(-\frac{\Delta G^\ddagger}{RT}\right)
$$

Because $\Delta G^\ddagger$ appears in the exponential, even a modest reduction in the barrier can produce a very large increase in reaction rate. **At $25^\circ\mathrm{C}$, lowering $\Delta G^\ddagger$ by approximately $5.7\ \mathrm{kJ,mol^{-1}}$ increases the rate about tenfold!**

Enzymes accelerate reactions by providing a pathway with a lower activation free-energy barrier. They do not change $\Delta G_{\mathrm{rxn}}$, the equilibrium constant, or the final equilibrium composition; they simply allow equilibrium to be reached faster.

(while the transition-state theory provides a framework for researchers to understand enzyme catalysis, it also has limitations that we won't get into)

  

## **2. Applications of Enzymes**

Industrial biocatalysis can use purified enzymes, immobilized enzymes, or whole microbial cells that contain the required enzymes within pathways.

* **MATERIALS AND POLYMERS**

  * **Acrylamide:** **Nitrile hydratase** adds water to acrylonitrile to produce acrylamide. Acrylamide is subsequently polymerized into polyacrylamide, which is widely used in water treatment, papermaking, mining, and oil recovery. This is one of the largest-scale applications of industrial biocatalysis.

* **FINE CHEMICALS AND FLAVORINGS**

  * **Vanillin:** Commercial “natural” vanillin can be produced from plant-derived ferulic acid using *Amycolatopsis* whole-cell fermentation. **Feruloyl-CoA synthetase (Fcs)** first activates ferulic acid, and **enoyl-CoA hydratase/aldolase (Ech)** then converts feruloyl-CoA into vanillin.

* **FRAGRANCES AND PEST CONTROL**

  * **Nootkatone:** In engineered yeast, **valencene synthase** produces valencene, a **cytochrome P450 valencene oxidase** and its **P450 reductase** convert valencene into nootkatol, and an **alcohol dehydrogenase** oxidizes nootkatol into nootkatone. Nootkatone is used as a grapefruit flavor and fragrance and as an insect-repellent or insecticidal ingredient.

* **PHARMACEUTICALS**

  * **Islatravir:** Merck and Codexis developed a three-step, nine-enzyme cascade for its synthesis. Five core enzymes were engineered: **galactose oxidase**, **pantothenate kinase**, **deoxyribose-5-phosphate aldolase (DERA)**, **phosphopentomutase**, and **purine nucleoside phosphorylase**. Together, they perform selective oxidation, phosphorylation, carbon–carbon bond formation, phosphate transfer, and attachment of the nucleobase. Four auxiliary enzymes recycle cofactors and drive the reactions forward. [Original islatravir study](https://www.science.org/doi/10.1126/science.aay8484)

* **FOOD**

  * **Aspartame:** **Thermolysin**, normally a protease, can also catalyze peptide-bond formation in the reverse direction. It couples protected L-aspartic acid with L-phenylalanine methyl ester to make an aspartame precursor, which is subsequently deprotected.

## **RHEA**
Rhea is an expert-curated **knowledgebase** of **chemical reactions catalyzed by enzymes.**

Enzyme-catalyzed and spontaneously occurring reactions are curated from **peer-reviewed literature and represented in a computationally tractable manner.**

## **4. Deep learning for Enzyme Design: CONDITIONAL GENERATION**

Unconditional PLMs (like ProtGPT2, as we saw in the previous notebook) sample sequences exploring vast regions of the protein space.

**Conditioning** language models: To exert more **control over the generation process**, these machine learning models can be tailored to **specific properties** such as:
* Protein family
* Function
* Target organism.

In practice, we do this by either 1) preprending conditional tags to the sequence, and letting the model to infer left-to-right relationships autoregressively or (such as ZymCTRL, Progen-1 or CTRL) 2) adding an encoder that captures the conditioning first and passes the information through cross-attention to the decoder (such as REXzyme, that we will see later today).

Conditional language models can be trained adding a control tag to the training data.

Conditional PLM were aplied after the development of the conditional transformer language (CTRL) in natural language.


### **ZymCTRL**
* Stands for Enzyme Control.

* Model developed in the lab to enable controlled protein generation.

* ZymCTRL is a conditional language model conditioned on EC labels of enzymes.

* Released [first](https://www.mlsb.io/papers_2022/ZymCTRL_a_conditional_language_model_for_the_controllable_generation_of_artificial_enzymes.pdf) in 2022, based on the CTRL Transformer Architecture. Decoder-only model.


We can **generate sequences for a user-defined EC class**.


This model was **experimentally validated** by generating **lactate dehydrogenases** for sequences 40-50% identical to natural ones.

The training dataset contains 37M sequences from UniProt database of sequences containing EC labels as control tags.

During training, the model learned the relationship between EC-labels and sequence. It was trained in an autoregressive objective, the model learns to predict the next token given the sequence context.

## **REXzyme**

* Stands for, Reaction-Enzyme.

* Model developed in the lab to enable controlled protein generation.

* REXzyme is a encoder-decoder language model (such as translation machine) that relates protein sequences to SMILE reactions.

We can generate sequences for a **user-defined reactions** that are ordered, globular, distant to nature, and that match their catalytic activity.

Training data: reactions - sequences

### **SMILES**

SMILES (Simplified Molecular Input Line Entry System ) is a specification in the form **describing the structure of chemical species as a line of text** using short ASCII strings.

* Atoms are represented by their atomic symbols
* Bonds:
  * Single bonds are represented are implicit.
  * Double bonds are represented as =
  * Triple bonds are represented as #
  * Aromatic bonds are represented by :
3. Branches are represented using ()
4. Rings are represented using numbers to indicate the binining and end of the ring.
5. Aromaticity is defined by using lowercase letters
6. Hidrogens are usually implicit but they can also be specified
7. Ions and charges are represented using square brackets, with charges indicated as + or -
8. Stochemistry can be also descrived:
  - chirality: @ anticlockwise, @@clockwise
  - double bond confifuration: / cis and \ trans isomerism

A single molecule has multiple posible SMILE strings. The larger the molecule, the higher the number of possible SMILES. Useful for data augmentation.
However, we can convert all the possibilities to one, by using canonical SMILES.

# **ZYMCTRL HANDS ON** 🤗 🟦 :

ZymCTRL is a model for controlling enzyme generation. It is most useful when we want to augment the repertoire of natural enzyme families. For example, we may have data about an enzyme family with attractive properties, such as defluorinases (which have the potential to break down PFAS), but the enzymes may be scarces (there are fewer than 10 well-known) and they lack the efficiency to be used industrially. With ZymCTRL, we can generate new variants and expand this family, with the potential that some of those variants may be more catalytically efficient or present broadened substrate scope.

This model can generate proteins for user-defined EC-classes.

```python
import torch
from transformers import GPT2LMHeadModel, AutoTokenizer
import os
from tqdm import tqdm
import math

def remove_characters(sequence, char_list):
    "This function removes special tokens used during training."
    columns = sequence.split('<sep>')
    seq = columns[1]
    for char in char_list:
        seq = seq.replace(char, '')
    return seq

def calculatePerplexity(input_ids,model,tokenizer):
    "This function computes perplexities for the generated sequences"
    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
    loss, logits = outputs[:2]
    return math.exp(loss)

def main(label, model,special_tokens,device,tokenizer):
    # Generating sequences
    input_ids = tokenizer.encode(label,return_tensors='pt').to(device)
    outputs = model.generate(
        input_ids,
        top_k=9, #tbd
        repetition_penalty=1.2,
        max_length=1024,
        eos_token_id=1,
        pad_token_id=0,
           do_sample=True,
           num_return_sequences=5) # Depending non your GPU, you'll be able to generate fewer or more sequences. This runs in an A40.

    # Check sequence sanity, ensure sequences are not-truncated.
    # The model will truncate sequences longer than the specified max_length (1024 above). We want to avoid those sequences.
    new_outputs = [ output for output in outputs if output[-1] == 0]
    if not new_outputs:
        print("not enough sequences with short lengths!!")

    # Compute perplexity for every generated sequence in the batch
    ppls = [(tokenizer.decode(output), calculatePerplexity(output, model, tokenizer)) for output in new_outputs ]

    # Sort the batch by perplexity, the lower the better
    ppls.sort(key=lambda i:i[1]) # duplicated sequences?

    # Final dictionary with the results
    sequences={}
    sequences[label] = [(remove_characters(x[0], special_tokens), x[1]) for x in ppls]

    return sequences

if __name__=='__main__':
    device = torch.device("cuda") # Replace with 'cpu' if you don't have a GPU - but it will be slow
    print('Reading pretrained model and tokenizer')
    tokenizer = AutoTokenizer.from_pretrained('AI4PD/ZymCTRL') # change to ZymCTRL location
    model = GPT2LMHeadModel.from_pretrained('AI4PD/ZymCTRL').to(device) # change to ZymCTRL location
    special_tokens = ['<start>', '<end>', '<|endoftext|>','<pad>',' ', '<sep>']

    # change to the appropriate EC classes
    labels=['3.5.5.1'] # nitrilases. You can put as many labels as you want.

    for label in tqdm(labels):
        # We'll run 100 batches per label. 20 sequences will be generated per batch.
        for i in range(0,10):
            sequences = main(label, model, special_tokens, device, tokenizer)
            for key,value in sequences.items():
                for index, val in enumerate(value):
                    # Sequences will be saved with the name of the label followed by the batch index,
                    # and the order of the sequence in that batch.
                    fn = open(f"{label}_{i}_{index}.fasta", "w")
                    fn.write(f'>{label}_{i}_{index}\t{val[1]}\n{val[0]}')
                    fn.close()
```
