## **1.1. Enzymes theory**

* Enzymes are **biological catalysts.**

* Enzymes are attractive because they can accelerate chemical reactions under mild conditions. Therefore, they can be used in industrial processes.


**BENEFITS** from using enzymes:
* **Highly specific**: higly selective, they can catalyze reactions for specific substrates.
* **Work under mild conditions**
* **Catalytic efficiency**: they can be coupled in cascades, then reducing the need of multiple chemical steps.
* **Eco-friendly:** They are sustainable as large-scale catalysts because enzymes are biodegradable, operate under mid reaction conditions, minimize the use of dangerous chemicals, waste, and energy consumption.They are also reusable.

**ENZYME STUCTURE:**
  * **Active site**: region where the substrate binds to produce the reaction. Higly specific for one substrate or a group of substrates.
  * **Cofactors and coenzymes**: additional molecules required to produce the  function:
    * Cofactors: inorganic ions (mg2+, Zn2+...)
    * Coenzymes: organic molecules (NAD+, FAD...)


**MECHANISM**:
* **Steps**:
1. Substrate binding: formation of enzyme-substrate complex.
2. Catalysis: Conversion of substrates into products.
3. Product Release: generation of the product.

Their active sites are highly specialized regions that lower the activation energy, accelerating chemical transformations by several orders of magnitude

* **Kinetics**:
  * Michalis-Menten: describes relationship between the substrate concentration and the reaction rate.
    * Vmax: maximum reaction rate
    * Km: substrate concentration at which the reaction rate is half Vmax. Low km indicates high substrate affinity.
    * Kcat: number of substrate molecules converted to product per enzyme per second.

## **RHEA**
Rhea is an expert-curated **knowledgebase** of **chemical reactions catalyzed by enzymes.**

Enzyme-catalyzed and spontaneously occurring reactions are curated from **peer-reviewed literature and represented in a computationally tractable manner.**

Rhea reactions are described independent of their biological location (subcellular location, tissue, organ, species) and are **chemically balanced** at pH 7.3.

Rhea is freely available and accessible in different formats.

It links the information to other resources:

## **4. Deep learning for Enzyme Design: CONDITIONAL GENERATION**

Unconditional PLMs (like ProtGPT2) sample sequences exploring vast regions of the protein space.

**Conditioning** language models: To exert more **control over the generation process**, these machine learning models can be tailored to **specific properties** such as:
* Protein family
* Function
* Target organism.

Conditional language models can be trained adding a control tag to the training data.

Conditional PLM were aplied after the development of the conditional transformer language (CTRL) in natural language.


### **ZymCTRL**
* Stands for, Enzyme Control.

* Model developed in the lab to enable controlled protein generation.

* ZymCTRL is a conditional language model conditioned on EC labels of enzymes.

* Released in 2022, based on the CTRL Transformer Architecutre. Decoder-only model.


We can **generate sequences for a user-defined EC class**.


This model was **experimentally validated** by generating **lactate dehydrogenases**, even for sequences only 40% similar to the natural ones.










Geraldene Munsamy et al. Conditional language models enable the efficient design of proficient enzymes, (2024).

The training dataset contains 37M sequences from UniProt database of sequences containing EC labels as control tags.

During training, the model learned the relationship between EC-labels and sequence. It was trained in an autoregressive objective, the model learns to predict the next token given the sequence context.

### **SMILES**

SMILES (Simplified Molecular Input Line Entry System ) is a specification in the form **describing the structure of chemical species as a line of text** using short ASCII strings.

* Atoms are represented by their atomic symbols
* Bonds:
  * Single bonds are represented are implicit.
  * Double bonds are represented as =
  * Triple bonds are represented as #
  * Aromatic bonds are represented by :
3. Branches are represented using ()
4. Rings are represented using numbers to indicate the binining and end of the ring.
5. Aromaticity is defined by using lowercase letters
6. Hidrogens are usually implicit but they can also be specified
7. Ions and charges are represented using square brackets, with charges indicated as + or -
8. Stochemistry can be also descrived:
  - chirality: @ anticlockwise, @@clockwise
  - double bond confifuration: / cis and \ trans isomerism



image.png


A single molecule has multiple posible SMILE strings. The larger the molecule, the higher the number of possible SMILES. Useful for data augmentation.
However, we can convert all the possibilities to one, by using canonical SMILES.

```python
import torch
from transformers import GPT2LMHeadModel, AutoTokenizer
import os
from tqdm import tqdm
import math

def remove_characters(sequence, char_list):
    "This function removes special tokens used during training."
    columns = sequence.split('<sep>')
    seq = columns[1]
    for char in char_list:
        seq = seq.replace(char, '')
    return seq

def calculatePerplexity(input_ids,model,tokenizer):
    "This function computes perplexities for the generated sequences"
    with torch.no_grad():
        outputs = model(input_ids, labels=input_ids)
    loss, logits = outputs[:2]
    return math.exp(loss)

def main(label, model,special_tokens,device,tokenizer):
    # Generating sequences
    input_ids = tokenizer.encode(label,return_tensors='pt').to(device)
    outputs = model.generate(
        input_ids,
        top_k=9, #tbd
        repetition_penalty=1.2,
        max_length=1024,
        eos_token_id=1,
        pad_token_id=0,
           do_sample=True,
           num_return_sequences=5) # Depending non your GPU, you'll be able to generate fewer or more sequences. This runs in an A40.

    # Check sequence sanity, ensure sequences are not-truncated.
    # The model will truncate sequences longer than the specified max_length (1024 above). We want to avoid those sequences.
    new_outputs = [ output for output in outputs if output[-1] == 0]
    if not new_outputs:
        print("not enough sequences with short lengths!!")

    # Compute perplexity for every generated sequence in the batch
    ppls = [(tokenizer.decode(output), calculatePerplexity(output, model, tokenizer)) for output in new_outputs ]

    # Sort the batch by perplexity, the lower the better
    ppls.sort(key=lambda i:i[1]) # duplicated sequences?

    # Final dictionary with the results
    sequences={}
    sequences[label] = [(remove_characters(x[0], special_tokens), x[1]) for x in ppls]

    return sequences

if __name__=='__main__':
    device = torch.device("cuda") # Replace with 'cpu' if you don't have a GPU - but it will be slow
    print('Reading pretrained model and tokenizer')
    tokenizer = AutoTokenizer.from_pretrained('AI4PD/ZymCTRL') # change to ZymCTRL location
    model = GPT2LMHeadModel.from_pretrained('AI4PD/ZymCTRL').to(device) # change to ZymCTRL location
    special_tokens = ['<start>', '<end>', '<|endoftext|>','<pad>',' ', '<sep>']

    # change to the appropriate EC classes
    labels=['3.5.5.1'] # nitrilases. You can put as many labels as you want.

    for label in tqdm(labels):
        # We'll run 100 batches per label. 20 sequences will be generated per batch.
        for i in range(0,10):
            sequences = main(label, model, special_tokens, device, tokenizer)
            for key,value in sequences.items():
                for index, val in enumerate(value):
                    # Sequences will be saved with the name of the label followed by the batch index,
                    # and the order of the sequence in that batch.
                    fn = open(f"{label}_{i}_{index}.fasta", "w")
                    fn.write(f'>{label}_{i}_{index}\t{val[1]}\n{val[0]}')
                    fn.close()
```
